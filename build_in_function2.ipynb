```python
def find_max_min(numbers):
    max_num = max(numbers)
    min_num = min(numbers)
    return max_num, min_num

nums = []
for i in range(5):
    num = int(input(f"Enter number {i+1}: "))
    nums.append(num)

maximum, minimum = find_max_min(nums)

print("Numbers entered:", nums)
print("Maximum number:", maximum)
print("Minimum number:", minimum)
```

```python
string = "To change the overall look of your document. To change the look available in the gallery"

words = string.lower().split()

word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

for word, count in word_count.items():
    print(word, ":", count)
```

```python
string = "Welcome to python Training"

vowels = "aeiouAEIOU"
count = 0

print("Vowels in the string:")

for char in string:
    if char in vowels:
        print(char, end=" ")
        count += 1

print("\nTotal number of vowels:", count)
```

```python
string = "P@#yn26at^&i5ve"

chars = 0
digits = 0
symbols = 0

for ch in string:
    if ch.isalpha():
        chars += 1
    elif ch.isdigit():
        digits += 1
    else:
        symbols += 1

print("Chars =", chars)
print("Digits =", digits)
print("Symbols =", symbols)
```

```python
string = "String and String Function"

words = string.split()
result = []

for word in words:
    if word not in result:
        result.append(word)

output = " ".join(result)
print(output)
```

```python
string = "Hell0 W0rld ! 123 * # welcome to pYtHoN"

upper = 0
lower = 0
number = 0
special = 0

for ch in string:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        number += 1
    else:
        special += 1

print("UpperCase :", upper)
print("LowerCase :", lower)
print("NumberCase :", number)
print("SpecialCase :", special)
```
