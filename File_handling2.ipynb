```python
file = open("data.txt","w+")
file.write("Welcome to python")
file.seek(0)
content=file.read()
print(content)
file.close()
```

```python
file = open("stud1.txt","a+")
file.write("\nname:ABC")
file.seek(0)
content=file.read()
print(content)
file.close()
```

```python
file = open("data.txt","w+")
file.write("Welcome to python")
file.seek(0)
file = open("data.txt", "r")

content = file.read()
words = content.split()
print("Total number of words in file:", len(words))

file.close()
```

```python
try:
    file = open("dat.txt", "r")
    content = file.read()
    print("File content:\n", content)
    file.close()
except FileNotFoundError:
    print("Error: The file does not exist.")
```
