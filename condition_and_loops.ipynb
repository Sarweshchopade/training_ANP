```python
year=int(input("Enter a year: "))
if (year%400 ==0) or (year%4 ==0 and year%100!=0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
c=int(input("Enter third number: "))
if a>=b and a>=c:
    print("Largest number is:", a)
elif b>=a and b>=c:
    print("Largest number is:", b)
else:
    print("Largest number is:", c)
```

```python
product_code=int(input("Enter product code (1/2/3): "))
amount=float(input("Enter order amount: "))
discount=0
if product_code==1:
    if amount>1000:
        discount=0.10*amount
elif product_code==2:
    if amount>100:
        discount=0.05*amount
elif product_code==3:
    if amount>500:
        discount=0.10*amount
else:
    print("Invalid product code")
net_amount=amount-discount
print("Net amount to be paid: Rs.", net_amount)
```

```python
distance=int(input("Enter distance in km: "))
if distance>=1 and distance<=50:
    fare=distance*8
elif distance>=51 and distance<=100:
    fare=distance*10
elif distance>100:
    fare=distance*12
else:
    fare=0
    print("Invalid distance")
print("Total fare: Rs.", fare)
```

```python
num=int(input("Enter a number: "))
rev=0
while num>0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
print("Reversed number:", rev)
```

```python
num=int(input("Enter a number: "))
temp=num
rev=0
while num>0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
if temp==rev:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

```python
num=int(input("Enter a number: "))
fact=1
i=1
while i<=num:
    fact=fact*i
    i=i+1
print("Factorial =", fact)
```

```python
total=0
while True:
    num=int(input("Enter a number: "))
    if num==0:
        break
    total=total+num
print("Sum of entered numbers:",total)
```

```python
s=input("Enter a string: ")
if s==s[::-1]:
    print("Palindrome string")
else:
    print("Not a palindrome string")
```

```python
num=int(input("Enter a number: "))
temp=num
sum=0
while temp>0:
    digit=temp%10
    sum=sum+(digit ** 3)
    temp=temp//10
if sum==num:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
a=0
b=1
print("Fibonacci series up to 50:")
while a<=50:
    print(a)
    a, b=b, a+b
```

```python
password=input("Enter password: ")
if (len(password)>=8 and
    any(char.isupper() for char in password) and
    any(char.islower() for char in password) and
    any(char.isdigit() for char in password)):
    print("Valid password")
else:
    print("Invalid password")
```
