```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

union_set = set1.union(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Union of sets:", union_set)
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

union_set = set1.intersection(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Union of sets:", union_set)
```
