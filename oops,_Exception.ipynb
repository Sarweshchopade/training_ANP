What is Object Oriented Programming? Explain class and object with a simple Python example.

--> Object Oriented Programming (OOP) is a programming approach where a program is designed using objects and classes.
It focuses on real-world entities and helps in code reusability, security, and maintainability.

Class:
A class is a blueprint or template used to create objects.

Object:
An object is an instance of a class. It represents a real-world entity.

Example:

```python
class Student:
    def display(self):
        print("This is a student class")
obj=Student()
obj.display()
```

Explain the role of the __init__ method in a class. When is it called?

--> The __init__ method is a constructor in Python.

It is automatically called when an object is created.

Example:

```python
class Student:
    def __init__(self, name, marks):
        self.name=name
        self.marks=marks
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
s1=Student("sarwesh", 85)
s1.display()
```

What is encapsulation? How does Python achieve data hiding? Explain with an example.

--> Encapsulation is an Object Oriented Programming (OOP) concept that means wrapping data (variables) and methods (functions) together into a single unit (class) and restricting direct access to the data.

Python achieves data hiding using access specifiers:

Public

Protected

Private

Example:


```python
class BankAccount:
    def __init__(self, name, balance):
        self.name=name
        self._account_type="Saving"
        self.__balance=balance
    def show_balance(self):
        print("Balance:", self.__balance)
    def deposit(self, amount):
        self.__balance+=amount
acc=BankAccount("Amit", 5000)
acc.show_balance()
acc.deposit(2000)
acc.show_balance()
```

What is inheritance? Write a Python program to show single inheritance.

--> Inheritance is an important concept of Object Oriented Programming (OOP) in which a child class (derived class) acquires the properties and methods of a parent class (base class).

Single Inheritance:

Single inheritance means a child class inherits from only one parent class.

Example:


```python
class Parent:
    def show_parent(self):
        print("This is the Parent class")
class Child(Parent):
    def show_child(self):
        print("This is the Child class")
obj=Child()
obj.show_parent()
obj.show_child()
```

What is an exception in Python? Why is exception handling required?

--> An exception in Python is an error that occurs during the execution of a program (runtime error) and disrupts the normal flow of the program.

Exception handling is required to handle runtime errors gracefully without stopping the entire program.

Example:

```python
try:
    a=int(input("Enter a number: "))
    b=int(input("Enter another number: "))
    print(a/b)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed")
except ValueError:
    print("Error: Please enter valid integers")
```

Write a Python program to handle division by zero and invalid input using exception handling.

-->

```python
try:
    a=int(input("Enter first number: "))
    b=int(input("Enter second number: "))
    result=a/b
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed")
except ValueError:
    print("Error: Please enter valid integer values")
finally:
    print("Program execution completed")
```
